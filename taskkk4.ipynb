```python
audio_file = "TASK 4 SAMPLE .mp4"
```

```python
output_audio = "extracted_audio.wav"

subprocess.run([
    "ffmpeg",
    "-i", audio_file,
    "-vn",
    "-acodec", "pcm_s16le",
    "-ar", "16000",
    "-ac", "1",
    output_audio
])
```

```python
whisper_model = whisper.load_model("small")  # use "base" if low VRAM
```

```python
result = whisper_model.transcribe(output_audio)

transcribed_text = result["text"]

print("🔹 Transcribed Text:\n")
print(transcribed_text)
```

```python
model_name = "unsloth/mistral-7b-bnb-4bit"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=model_name,
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True
)

FastLanguageModel.for_inference(model)
```

```python
def generate_reasoning(text):

    prompt = f"""
You are a logical reasoning assistant.
Analyze the following query and provide a clear step-by-step answer.

Query:
{text}

Answer:
"""

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    outputs = model.generate(
        **inputs,
        max_new_tokens=200,
        temperature=0.3,
        do_sample=True
    )

    response = tokenizer.decode(outputs[0], skip_special_tokens=True)

    return response
```

```python
final_answer = generate_reasoning(transcribed_text)

print("\n Final Reasoned Answer:\n")
print(final_answer)
```
